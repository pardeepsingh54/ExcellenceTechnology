```python
# 21. Create a function count_digits() that accepts an integer and returns the total number of digits.

def count_digits(n):
    count = 0

    while n>0 :
        n = n//10
        count = count+1
    return count
count_digits(1234)
# count_digits(12)
```

```python
# 22. Create a function sum_of_digits() that accepts an integer and returns the sum of its digits

def sum_of_digits(n):
    total = 0

    while n>0:
        digit = n % 10
        total = total+digit
        n = n // 10
    return total
sum_of_digits(12345)
```

```python
# 23. Create a function largest_in_list() that accepts a list of numbers and returns the largest element.

def largest_in_list(number):
    largest = number[0]

    for num in number:
        if num > largest:
            largest = num
    return largest
print(largest_in_list([5,3,4,6,7,8,9,10]))
```

```python
# 27. Create a function fibonacci() that accepts n and prints/returns the first n Fibonacci numbers.

def fibonacci(n):
    a=0
    b=1
    series = []
    for i in range(n):
        series.append(a)
        a,b = b,a+b
    return series
fibonacci(10)
```

```python
# 28. Create a function is_armstrong() that checks whether a number is an Armstrong number.

def is_armstrong(num):
    num_str = str(num)
    power =  len(num_str)
    total = 0
    for digit_str in num_str:
        digit = int(digit_str)
        total = total + (digit**power)

    if total == num:
        return True
    else:
        return False
is_armstrong(370)
is_armstrong(153)
is_armstrong(371)
```

```python
# 31. Write a function that accepts a list and returns the second largest element without using sort().
def second_largest(nums):

    largest = max(nums)

    new_list = []

    for num in nums:
        if num!=largest:
            new_list.append(num)

    second = max(new_list)
    
    return second

second_largest([12,332,556,744,221,100])
```

```python
# 32. Write a function that accepts a sentence and returns the word having the maximum length.
def longest_word(sentence):

    words = sentence.split()

    if not words:
        return ""

    longest = words[0]
    for word in words:
        if len(word) > len(longest):
            longest = word
    return longest
longest_word("my name is pardeep")
```

```python
# 33. Write a function that accepts a list of integers and returns separate lists of even and odd numbers.

def list_of_integers(num):
    even_list = []
    odd_list = []
    for i in num:
        if i%2==0:
            even_list.append(i)
        elif i%2!=0:
            odd_list.append(i)

    return even_list,odd_list
evenn ,oddd =list_of_integers([1,2,3,4,5,6,7,8,9])
print("Even : ",evenn)
print("Odd : ",oddd)
```
