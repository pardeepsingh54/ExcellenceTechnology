Q1. Define a Set in Python. Write any four characteristics of sets.
Ans: A set in python is an unordered collections of unique elements.

Four characteristics of sets:

Unordered : Elements in a set have no fiexed order.
Unique : A set does not allow duplicate elements.
Mutable : Elements can be added or remove from a set.
Heterogeneous : A set can contain diffrent data types, such as integer, string and float.

Q9. Explain the difference between remove() and discard() with examples.

```python
#Both remove() and discard() are used to remove an element from a set, but there is one important differnce:
#remove() : remove() removes the specified element, if the element does not exist, it gives a keyerror.
my_set = {12,21,34,43,56,65}

my_set.remove(60)
print(my_set)

#discard() : discard removes the specified element, if the element does not exise, it does nothing and giving no error.
my_set.discard(60)
print(my_set)
```
