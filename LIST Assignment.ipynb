Q3. What is the difference between append() and extend()?

```python
#append() : Adds its argument as a single element to the end of the list.
#extend() : itrates over its argument and adds each item individually to the end of the list.
list1 = [1,2,3,4,5]
list1.append(6)     #append()
print(list1)      

list1.extend([7,8,9,10])    #extend()
print(list1)
```

Q5. What does the len() function do?

```python
# The len() function is a built-in python funcion that returns the total number of items contained in an object.
# It works with sequences like string, list, and tuples, as well as collections like sets and dictionaries. Calling 
# len() on an empty object return 0, while passing an unsupported type (like an integer) raises a TypeError.

number_list = [90,80,70,60,50,40,30,20,10]

print(len(number_list))    
```
