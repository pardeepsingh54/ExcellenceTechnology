Q2. WAP to create two frozenset objects and perform:
    Union,
    Intersection,
    Difference

```python
ob1 = frozenset([1,2,3,4,5])
ob2 = frozenset([5,4,3,6,7])

print(ob1.union(ob2))

print(ob1.intersection(ob2))

print(ob1.difference(ob2))
```

Q4. WAP ro demonstrate that frozenset is immutable by trying to add or remove an element.

```python
fs = frozenset({12,33,44,55,66})
fs.add(111)         #'frozenset' object has no attribute 'remove'
print(fs)
fs.remove(12)
print(fs)           #'frozenset' object has no attribute 'remove'
```

Q9. WAP to check whether one frozenset is a subset of another frozenset.

```python
fs = frozenset([1,2])
fs1 = frozenset([1,2,3,4])

if fs.issubset(fs1):
    print("fs is a subset of fs1")
else:
    print("fs is not a subset of fs1")
```
