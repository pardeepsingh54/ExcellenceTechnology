Section 8: Challenge Questions

1.	Why is dictionary faster than list for searching?

Ans. A Dictionary is faster than a list for searching because a dictionary uses hashing to store and locate its keys. When we search for a key , python 
can quickly determine where the key is stored without checking every element one by one.

In contrast, a list usually searches elements sequentially, starting from the first element until it find the required item.

Therefore, dictionary searching has an average time complexity of O(1), while list searching has a time complexity of O(n). This makes dictionaries more 
efficient and faster than lists for searching, especilly when dealing with large amounts of data.

2.	Can a dictionary store another dictionary as a value? Explain.

Ans. Yes dictionary can store another dictionary as value. This is known as nested dictionary. It allows us to store related and structure data inside another dictionary. The inner dictionary can have its own keys and values, making it useful for representing complex or hieraarchical data.

3.	Why list cannot be used as a dictionary key?

Ans. A list cannot be used as a dicyionary key because lists are mutable, meaning their elements can be changed after the creation. Dictionary keys must be immutable and hashable so that pyhton can calculate an maintain their hash value.

Since a list can change, its hash value cannot remain fixed. Therfore, Python does not allow lists to be used as dictionary keys.
